```python
def convert_rgb_to_grayscale(r: int, g: int, b: int) -> int:
    #include luminosity formula (eye weights: R=0.299, G=0.587, B=0.114)
    luminosity = 0.299 * r + 0.587 * g + 0.114 * b 

    # Round to int (pixel 0-255)
    grayscale_val = int(round(luminosity))
    return grayscale_val

# Test with sample values 
sample_r, sample_g, sample_b = 180, 120, 60
result = convert_rgb_to_grayscale(sample_r, sample_g, sample_b)

print(f"Input RGB: ({sample_r}, {sample_g}, {sample_b})")
print(f"Calculated Grayscale Value: {result}")
assert result == 131, f"Expected 131, but got {result}"     # To verify
print("Manual Trace Passed Successfully!")
```

```python
import numpy as np

# Python implementation  (vectorized with numpy)
def adjust_brightness_contrast(image: np.ndarray, alpha: float = 1.0, beta: float =0.0) ->np.ndarray:
    """Applies linear point processing 0 = alpha * I + beta"""
    # Convert to float32 to avoid overflow during calculation
    image_float = image.astype(np.float32)

    # Apply brightness/contrast formula on whole array at once 
    adjusted = (alpha * image_float) + beta

    # Clip values to valid image range  [0, 255]
    clipped = np.clip(adjusted, 0, 255)

    # Convert back to uint8 for image format
    return clipped.astype(np.uint8)

# Manual trace data for verification
test_matrix = np.array([[100, 150],
                        [200, 50]],  dtype=np.uint8)
# Apply alpha = 1.5 for contrast, beta = -20 for brightness
output_matrix = adjust_brightness_contrast(test_matrix, alpha = 1.5, beta = -20.0)

print("Original Matrix:\n", test_matrix)
print("\nTransformed Matrix:\n", output_matrix)

#Expected output: [[130, 205], [255, 55]]
expected_matrix = np.array([[130, 205],
                            [255, 55]], dtype=np.uint8)
# To verify result
np.testing.assert_array_equal(output_matrix, expected_matrix)
print("\nVectorized Array Trace Passed Successfully!")
```

```python
import numpy as np
def apply_mean_filter_3x3(neighborhood: np.ndarray) -> int:
    """Computes the mean of a 3x3 matrix block and returns the rounded integer center value."""
    # Step 1: sum all 9 elements in the 3x3 matrix
    total_sum = np.sum(neighborhood)     #10+20+10+30+50+30+10+20+10 = 190

    #Step 2: Divide sum by 9 to get mean (average)
    mean_value = total_sum / 9.0  #190 / 9 = 21.11

    # Step 3: Round to nearest integer and return as int 
    return int(round(mean_value))                 # round(21.11) = 21

# Test implementation
sample_3x3 = np.array([[10, 20, 10],
                       [30, 50, 30],
                       [10, 20, 10]], dtype=np.uint8)
# Run and verify
result = apply_mean_filter_3x3(sample_3x3)
print("Mean Filter Center Pixel Result", result)
assert result == 21, f"Expected 21, got {result}"
print("Test Passed Successfully!")
```
